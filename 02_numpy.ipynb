### Numpy Array Operations

```python
import numpy as np
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Add:", a + b)
print("Multiply:", a * b)
print("Sum:", np.sum(a))
```

### Basic Slicing


```python
arr = np.arange(11)
print("Basic Slicing: ", arr[2:7])
print("With Step: ",arr[1:8:2])
print("Negative: ",arr[-3])
```

```python
arr_2d = np.array([[1,2,3],
                   [3,5,2],
                   [7,5,0]])
print("specific element: ",arr_2d[1,2])

#Target complete Row
print("Entire Row: ",arr_2d[1])

#Target complete Column
print("Entire Column: ",arr_2d[:, 1])
```

### Sorting

```python
unsorted = np.array([0,2,3,1,14,2,1,4,5,2,9,8])
print("Sorted array: ", np.sort(unsorted))

#Sort single Column
arr_2d[:,2] = np.sort(arr_2d[:,2])
print("Sort Single Column: ", arr_2d) 
#Sort 2D Array
print("2D Array Sorted by Column: ", np.sort(arr_2d,axis=0))
```

### FIltering and Masks


```python
marks = np.array([45,78,32,90,56,88,24,67])
print("Marks Greater or Equal to 50: ",marks[marks>=50]) #Filter the Numbers from the Array 

mask = marks < 50
print("Marks less than 50: ",marks[mask]) # store an expression in mask and give to marks array subscritp 

idx = [1,3,6]
print(marks[idx]) # it gives us the values at following indexes


```

### np.where and deleting elements

```python
temperature = np.array([29,35,41,30,38,25,43])
where = np.where(temperature>35)
print(where) #give the indexes where condition is TRUE


where = np.where(temperature>35,"HOT","NORMAL") # if condition TRUE : Replace the INDEX value With HOT, ELSE: REPLACE IT WITH Normal.
print(where)

```

### Working with real world data 

```python
# Columns: temperature, humidity, soil moisture
data = np.array([
    [30, 60, 25],
    [36, 45, 12],
    [28, 70, 35],
    [40, 35, 8],
    [32, 55, 18],
    [34, 65, 14]
])

print(data.shape)
print(data)

#Filter dry plots
dry_plots =np.where(data[:,2]<15)
print(dry_plots)
print(data[dry_plots])

# Locate hot plots
temp=np.where(data[:,0]>35)
print(temp)

# Generate warnings, Create a status array: 1 when soil moisture is below 15, otherwise 0.
warning = np.where(data[:,2]<15,1,0)
print(warning)

# Select specific plots
fancy_indexing = [1,3,5]
print("Plots: \n",data[fancy_indexing])

# Remove a record
data = np.delete(data,2,axis = 0)
print(data)

```
