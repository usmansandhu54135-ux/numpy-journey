### Start Numpy

```python
import numpy as np
```

### Creating Array from list


```python
arr_1d = np.array([1,2,3,4,5])
print("1D array: ",arr_1d)

arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D array: ",arr_2d)
```

### List vs Numpy array


```python
py_list = [1,2,3]
print("python list multiplication:" ,py_list*2)

arr_1d = np.array([1,2,3,4,5]) #element wise multipliction
print("Python array multiplication: ",arr_1d*2)

import time
start = time.time()
py_list1 = [i*2 for i in range(1000000)]
print("\n List Operation time: ", time.time()-start)

start = time.time()
array = np.arange(1000000)*2
print("\n Array Operation time: ", time.time()-start)
```

### Creating Array from Scratch

```python
zeros = np.zeros((3,4))
print("Zeros Array: \n", zeros)

ones = np.ones((3,4))
print("Ones Array: \n", ones)

full = np.full((2,2),7) # For same element in every index
print("Full Array: \n", full)

random = np.random.random((2,3))
print("Random Array: \n", random)

sequence = np.arange(0,11,2)
print("Sequence Array: \n", sequence)
```

### Vector,Matrix and Tensor

```python
vector = np.array([1,2,3,4])
print("vector", vector)

matrix = np.array([[1,2,3],[4,5,6]])
print("Matrix", matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("Tensor", tensor)
```

### Array Properties

```python
arr = np.array([[1,2,3],
                [4,5,6]])

print("Shape: ", arr.shape) #Gives Number of Rows and Columns
print("Dimension: ", arr.ndim) #Gives Number of Dimensions of Array
print("Size: ", arr.size) #Total Number of Elements of Array
print("DType: ", arr.dtype) #Give Data type
```

### Array Reshaping

```python
arr = np.arange(12)
print("Original Array: ", arr)

#reshape
reshape = arr.reshape((3,4))
print("Reshaped Array: ", reshape)

# flattend
flattend = reshape.flatten()
print("Flattend Array: ", flattend)

#ravel (return view, instead of copy)
raveled = reshape.ravel()
print("Raveled Array: ", raveled)

#Transpose
transpose = reshape.T
print("Transpose: \n", transpose)

```
